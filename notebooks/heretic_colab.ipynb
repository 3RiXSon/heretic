# Heretic on free Google Colab GPU

Runs [Heretic](https://github.com/p-e-w/heretic) on Colab’s free **T4** GPU.

**Before you start:** Runtime → Change runtime type → Hardware accelerator → **T4 GPU**.

Free Colab sessions can disconnect; keep `n_trials` low for a first pass. Use `bnb_4bit` so a small instruct model fits in ~15 GB VRAM.

```python
# Install Heretic (pulls CPU torch from PyPI indexes by default).
# Reinstall a CUDA wheel that matches Colab after that.
%pip install -q -U heretic-llm
%pip install -q --upgrade --force-reinstall torch torchvision --index-url https://download.pytorch.org/whl/cu124
```

```python
from pathlib import Path

config = """
# Free-Colab profile: 4-bit quant + short trial budget + non-interactive save.
quantization = "bnb_4bit"
device_map = "auto"
batch_size = 0
max_batch_size = 32
n_trials = 20
n_startup_trials = 8
study_checkpoint_dir = "checkpoints"
checkpoint_action = "restart"
trial_index = 0
model_action = "save"
save_directory = "heretic-output"
""".strip()

Path("config.toml").write_text(config + "\n", encoding="utf-8")
print("Wrote config.toml")
```
